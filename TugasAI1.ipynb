**2. Prediksi Ketersediaan Lowongan Kerja untuk 2025-2026 menggunakan Algoritma SARIMA (Seasonal ARIMA)**

Model ini digunakan untuk memprediksi ketersediaan lowongan kerja untuk tahun 2025-2026, dengan mempertimbangkan faktor musiman dalam data historis.

```python
# Menghitung jumlah lowongan kerja terdaftar dengan menjumlahkan lowongan kerja laki-laki dan perempuan
df['Lowongan Kerja Terdaftar - Jumlah'] = df['Lowongan Kerja Terdaftar - Laki-Laki'] + df['Lowongan Kerja Terdaftar - Perempuan']

# Membangun model SARIMA
model_sarima = SARIMAX(df['Lowongan Kerja Terdaftar - Jumlah'], order=(1, 1, 1), seasonal_order=(1, 1, 1, 12))
model_sarima_fit = model_sarima.fit()

# Melakukan prediksi untuk tahun 2025-2026
predictions_sarima = model_sarima_fit.forecast(steps=2)

# Membuat DataFrame baru untuk menyimpan hasil prediksi
predicted_years = pd.DataFrame({'Tahun': ['2025', '2026'], 'Prediksi': predictions_sarima})

# Visualisasi hasil dengan menggunakan plot garis
plt.figure(figsize=(10, 6))
# Plot data sebenarnya
plt.plot(df['Tahun'], df['Lowongan Kerja Terdaftar - Jumlah'], marker='o', color='purple', label='Data Sebenarnya')
# Plot hasil prediksi
plt.plot(predicted_years['Tahun'], predicted_years['Prediksi'], marker='o', color='orange', linestyle='--', label='Prediksi SARIMA')
# Menambahkan judul dan label
plt.title('Prediksi Ketersediaan Lowongan Kerja 2025-2026')
plt.ylabel('Jumlah Lowongan Kerja')
plt.xlabel('Tahun')
# Memutar label pada sumbu x untuk meningkatkan keterbacaan
plt.xticks(rotation=45)
# Menambahkan legenda
plt.legend()
# Menambahkan grid untuk meningkatkan keterbacaan
plt.grid()
# Menyimpan hasil visualisasi sebagai file PNG
plt.savefig('prediksi_lowongan_kerja_sarima.png')
# Menampilkan plot
plt.show()
```
